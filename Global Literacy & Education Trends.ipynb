```python
# Import pandas
import pandas as pd

# Load all CSV files
df_adult = pd.read_csv(r"C:\Users\HeY!\Project_2\adult_literacy.csv")
df_youth = pd.read_csv(r"C:\Users\HeY!\Project_2\youth_literacy.csv")
df_illiteracy = pd.read_csv(r"C:\Users\HeY!\Project_2\illiteracy_population.csv")
df_gdp = pd.read_csv(r"C:\Users\HeY!\Project_2\gdp.csv")
df_school = pd.read_csv(r"C:\Users\HeY!\Project_2\schooling.csv")

# Check data
df_adult.head()
```

```python
# Rename columns for clarity
df_adult.rename(columns={
    "entity": "country",
    "Year": "year",
    "adult_literacy_rate__population_15plus_years__both_sexes__pct__lr_ag15t99": "adult_literacy"
}, inplace=True)

# Drop null values
df_adult = df_adult.dropna().drop_duplicates()

df_adult = df_adult[(df_adult['year'] >= 1990) & (df_adult['year'] <= 2026)]

df_adult.head()
```

```python
df_youth.head()
```

```python
# Rename columns for clarity
df_youth.rename(columns={
    "entity": "country",
    "Year": "year",
    "youth_literacy_rate__population_15_24_years__male__pct__lr_ag15t24_m": "youth_male",
    "youth_literacy_rate__population_15_24_years__female__pct__lr_ag15t24_f": "youth_female"
}, inplace=True)

# Drop null values
df_youth = df_youth.dropna().drop_duplicates()

df_youth = df_youth[(df_youth['year'] >= 1990) & (df_youth['year'] <= 2026)]

df_youth.head()
```

```python
df_illiteracy.head()
```

```python
# Rename columns for clarity
df_illiteracy.rename(columns={
    "entity": "country",
    "Year": "year",
    "youth_literacy_rate__population_15_24_years__male__pct__lr_ag15t24_m": "illiteracy_male",
    "youth_literacy_rate__population_15_24_years__female__pct__lr_ag15t24_f": "illiteracy_female"
}, inplace=True)

# Drop null values
df_illiteracy = df_illiteracy.dropna().drop_duplicates()

df_illiteracy = df_illiteracy[(df_illiteracy['year'] >= 1990) & (df_illiteracy['year'] <= 2026)]

df_illiteracy.head()
```

```python
df_gdp.head()
```

```python
# Rename columns for clarity
df_gdp.rename(columns={
    "entity": "country",
    "Year": "year",
    "ny_gdp_pcap_pp_kd": "gdp"
}, inplace=True)

# Drop null values
df_gdp = df_gdp.dropna().drop_duplicates()

df_gdp = df_gdp[(df_gdp['year'] >= 1990) & (df_gdp['year'] <= 2026)]

df_gdp.head()
```

```python
df_school.head()
```

```python
# Rename columns for clarity
df_school.rename(columns={
    "entity": "country",
    "Year": "year",
    "mf_youth_and_adults__15_64_years__average_years_of_education": "schooling"
}, inplace=True)

# Drop null values
df_school= df_school.dropna().drop_duplicates()

df_school = df_school[(df_school['year'] >= 1990) & (df_school['year'] <= 2026)]

df_school.head()
```

```python
# Merge adult + youth literacy
df_literacy = pd.merge(df_adult, df_youth, on=["country", "year"], how="inner")

# Merge GDP + schooling
df_gdp_school = pd.merge(df_gdp, df_school, on=["country", "year"], how="inner")
```

```python
#Literacy Gender Gap
df_literacy["gender_gap"] = (
    df_literacy["youth_male"] - df_literacy["youth_female"]
)
```

```python
#GDP per Schooling Year
df_gdp_school["gdp_per_school_year"] = (
    df_gdp_school["gdp"] /
    df_gdp_school["year"]
)
```

```python
#Education Index
# Merge literacy + schooling for this
df_temp = pd.merge(df_literacy, df_gdp_school, on=["country", "year"], how="inner")

# Create index
df_temp["education_index"] = (
    df_temp["adult_literacy"] * 0.5 +
    df_temp["year"] * 5
)
```

```python
#Youth Literacy Average
# Create average youth literacy
df_literacy["youth_avg"] = (
    df_literacy["youth_male"] + df_literacy["youth_female"]
) / 2
```

```python
#Literacy Growth Rate
# Sort data first (VERY IMPORTANT)
df_literacy = df_literacy.sort_values(by=["country", "year"])

# Calculate growth rate
df_literacy["literacy_growth"] = df_literacy.groupby("country")["adult_literacy"].pct_change() * 100
```

```python
# Combine adult and youth literacy
df_literacy["literacy_index"] = (
    df_literacy["adult_literacy"] * 0.6 +
    df_literacy["youth_avg"] * 0.4
)
```

```python
# Categorize literacy level
df_literacy["literacy_level"] = pd.cut(
    df_literacy["adult_literacy"],
    bins=[0, 60, 80, 100],
    labels=["Low", "Medium", "High"]
)
```

```python
# Difference between youth and adult literacy
df_literacy["generation_gap"] = (
    df_literacy["youth_avg"] - df_literacy["adult_literacy"]
)
#Positive → education improving in new generation
#Negative → decline (rare but powerful insight)
```

```python
df_illiteracy["female_to_male_ratio"] = (
    df_illiteracy["illiteracy_female"] /
    df_illiteracy["illiteracy_male"]
)
#1 → women more affected

#<1 → men more affected
```

```python
# Total illiterate population
df_illiteracy["illiterate_total"] = (
    df_illiteracy["illiteracy_male"] +
    df_illiteracy["illiteracy_female"]
)
```

```python
# Drop columns without error if not present
df_literacy = df_literacy.drop(columns=["code_x", "code_y", "owid_region"], errors="ignore")
```

```python
# Fill missing values with 0 for selected columns
df_literacy[["youth_avg", "gender_gap", "literacy_growth"]] = (
    df_literacy[["youth_avg", "gender_gap", "literacy_growth"]].fillna(0)
)
```

```python
# Import sqlite3 (built-in database)
import sqlite3

# Create connection (this creates a file database)
conn = sqlite3.connect("literacy.db")

# Create cursor to execute SQL
cursor = conn.cursor()

print("Database connected ✅")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS literacy_rates (
    country TEXT,
    year INTEGER,
    adult_literacy REAL,
    youth_male REAL,
    youth_female REAL
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS illiteracy_population (
    country TEXT,
    year INTEGER,
    illiteracy_male REAL,
    illiteracy_female REAL,
    illiterate_total REAL
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS gdp_schooling (
    country TEXT,
    year INTEGER,
    gdp REAL
    
)
""")
```

```python
# Insert data into literacy table
df_literacy[["country","year","adult_literacy","youth_male","youth_female"]].to_sql(
    "literacy_rates",
    conn,
    if_exists="replace",
    index=False
)
```

```python
df_illiteracy[["country","year","illiteracy_male","illiteracy_female","illiterate_total"]].to_sql(
    "illiteracy_population",
    conn,
    if_exists="replace",
    index=False
)
```

```python
df_gdp_school[["country","year","gdp"]].to_sql(
    "gdp_schooling",
    conn,
    if_exists="replace",
    index=False
)
```

```python
#QUERY 2: Female youth literacy < 80%
query = """
SELECT country, year, youth_female
FROM literacy_rates
WHERE youth_female < 80
ORDER BY country, year;
"""

pd.read_sql(query, conn)
```

```python
#3 Average adult literacy per continent (owid region)
query = """
SELECT year,
       COUNT(country) as country_count,
       ROUND(AVG(adult_literacy), 2) as avg_literacy
FROM literacy_rates
GROUP BY year
HAVING COUNT(country) > 20
ORDER BY year;
"""

pd.read_sql(query, conn)
```
